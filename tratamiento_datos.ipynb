TRATAMIENTO DE DATOS

```python
import pandas as pd
#cargando el archivo y mostrando cicno filas
df_original = pd.read_csv("base_ventas.csv", sep=";")
df_original.head()
```

```python
#duplicando el dataframe
df = df_original.copy()
```

```python
#cambiando el nombre de las columnas mpara estandarizar el nombre de las mismas (mayusculas primero y luegoi minusculas)
#imprimiendo los nombres originales
df.columns
#copiamos, estandarizamos y la modificacion la asignamos a una variable
nombre_columnas = ['matricula_funcionario', 'nombre_funcionario', 'cargo', 'codigo_tienda',
       'nombre_tienda', 'codigo_producto', 'descripcion_producto', 'categoria',
       'precio_costo', 'valor_unitario', 'cantidad', 'comision', 'dt_venta',
       'dt_entrega']

#cambiamos el nombre de las columnas con las nuevas
df.columns = nombre_columnas
#verificamos que se haya hecho el cambio
df.head()
```

CAMBIANDO EL TIPO DE DATO

```python
#analizamos el tipo de dato de todas las columnas
df.info()
```

```python
#asi se cambia el tipo de dato
df["matricula_funcionario"] = df["matricula_funcionario"].astype("str")
#o tambien así
df = df.astype({
    "codigo_producto": "str"
})
```

```python
#en el df hay columnas que tienen comas como separador decimal y eso no está bien, ya que se necesitan hacer calculos con las columnas hay que reemplazar las comas por puntos
df["precio_costo"] = df["precio_costo"].str.replace(",",".")
df["valor_unitario"] = df["valor_unitario"].str.replace(",",".")
```

```python
#convertimos a float para manejar decimales
df = df.astype({
    "precio_costo": "float",
    "valor_unitario": "float"
})

#verificamos que se haya hecho el cambio}
df.info()
```

```python
#cambiando la fecha a tipo date
df["dt_venta"] = pd.to_datetime(df["dt_venta"], format="%d/%m/%y %H:%M")
df["dt_entrega"] = pd.to_datetime(df["dt_entrega"], format="%d/%m/%y %H:%M")

df.info()
```

```python
#consultando partes de la fecha
df["dt_entrega"].dt.year
df["dt_entrega"].dt.month
df["dt_entrega"].dt.hour
```

TRTANDO DATOS FALTANTES

```python
#mostrando las columnas que tienen valores faltantes
df.isna().sum()
```

```python
#rellenando los valores faltantes
df.fillna(0).head()
#se recomienda hacerlo asi: df["columna donde hay valores faltantes"].fillna(0) para que solo afecte esa columna
#y con este df["columna donde hay valores faltantes"] = df["columna donde hay valores faltantes"].fillna(0) se hace el cambio definitivo
```

```python
#metodo ffill()
df["comision"].ffill()
```

```python
#metodo bfill()
df["comision"].bfill()
```

ELIMINANDO Y AGREGANDO COLUMNAS

```python
#eliminado una columna
df.drop(columns=["comision"])
#eliminado una columna de forma definitiva
#df.drop(columns=["comision"], inplace=True)
```

```python
#Ceando columnas calculadas
df["valor_total"] = df["valor_unitario"] * df["cantidad"]

df.head()
```

Ordenando el dataframe

```python
#para ordenar una sola comumna
df.sort_values("descripcion_producto", ascending=True)
#para ordenar dos o mas columnas. El inplace es para hacer el cambio permanente
#df.sort_values(["descripcion_producto", "nombre_tienda"], ascending=[True, True], inplace=True)
```

```python
#haciendo un reset del index. El drop significa que va a eliminar la columna vieja del indice
df.reset_index(drop=True)
#df.reset_index(drop=True, inplace=True) para hacer el cambio permanente
```

```python
#cambiar la columna que fungirá como indice. inplace es para haecr los cambios permanentes
df.set_index("dt_venta", inplace=True)
```

EXPORTANDO UN NUEVO DATASET

```python
#xportando a csv
df.to_csv("ventas_nuevo.csv")

#para que no exporte el index de pandas
df.to_csv("ventas_nuevo.csv", index=False)
```
